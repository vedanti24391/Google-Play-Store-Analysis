```python
apps['Reviews'] = pd.to_numeric(apps['Reviews'],errors='coerce')
```

```python
apps['Rating'] = pd.to_numeric(apps['Rating'],errors='coerce')
```

```python
apps['Installs'] = apps['Installs'].astype(str)
apps['Installs'] = apps['Installs'].str.replace(',','',regex=False)
apps['Installs'] = apps['Installs'].str.replace(',','',regex=False)
apps['Installs'] = pd.to_numeric(apps['Installs'],errors='coerce')
```

```python
apps['Price'] = apps['Price'].astype(str)
apps['Price'] = apps['Price'].str.replace('$','',regex=False)
apps['Price'] = pd.to_numeric(apps['Price'],errors='coerce')
```

```python
#handle missing values
apps.isnull().sum()
```

```python
apps["Rating"] = apps["Rating"].fillna(apps["Rating"].median())
```

```python
#duplicates
apps.duplicated().sum()
```

```python
#remove duplicates
apps = apps.drop_duplicates()
```

```python
#convert dates
apps['Last Updated'] = pd.to_datetime(apps['Last Updated'],errors = 'coerce')
```

```python
#handle size coloumn
def convert_size(Size):
    if pd.isna(Size):
        return np.nan

    if Size =="Varies with device":
        return np.nan
    
    Size = str(Size)

    if Size.endswith("M"):
        return float(Size[:1])

    elif Size.endswith("k"):
        return float(Size[:1])/1024

        return np.nan
apps["Size"] = apps["Size"].apply(convert_size)
```
