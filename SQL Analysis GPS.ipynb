```python
#App Rating Classification Using CASE Statement
query = """
SELECT App,
Rating,
CASE
WHEN Rating>=4.5 THEN 'Excellent'
WHEN Rating>=4 THEN 'Good'
ELSE 'Needs Improvement'
END AS Rating_Level
FROM apps;
"""
pd.read_sql(query, conn)
```

Excellent apps represent high customer satisfaction and are more likely to receive positive reviews and higher downloads.
Good apps perform well but may benefit from feature enhancements or bug fixes to achieve excellent ratings.
Needs Improvement apps may require immediate attention to improve user experience, fix issues, and increase customer satisfaction.

```python
#App Type Classification (Free vs Paid) Using CASE Statement
query = """
SELECT App,
Price,
CASE
WHEN Price=0 THEN 'Free'
ELSE 'Paid'
END AS App_Type
FROM apps;
"""
pd.read_sql(query, conn)
```

Free apps generally attract more users because there is no cost to download, helping businesses increase their user base.
Paid apps generate direct revenue but require users to perceive enough value to purchase them.
Categorizing apps as Free or Paid helps businesses compare user behavior, downloads, ratings, and revenue strategies across different pricing models.

```python
#Rank Apps by Rating
query ="""
SELECT App,
Rating,
RANK()OVER(ORDER BY Rating DESC) AS Rank_No
FROM apps;
"""
pd.read_sql(query,conn)
```

Ranking helps identify the top-performing apps based on user ratings.
Businesses can analyze highly ranked apps to understand what contributes to customer satisfaction and success.
Apps with lower ranks can be prioritized for improvements to increase user ratings.

```python
#Rank Apps Within Each Category
query="""
SELECT App,
Category,
Rating,
ROW_NUMBER()OVER(
PARTITION BY Category
ORDER BY Rating DESC
)AS Rank_In_Category
FROM apps;
"""

pd.read_sql(query,conn)

```

Ranking restarts from 1 for every new category because of the PARTITION BY Category clause.
Unlike RANK(), ROW_NUMBER() assigns a unique rank to every app, even if multiple apps have the same rating.
This query helps identify the top-performing apps in each category rather than across the entire Play Store.
Businesses can compare the best apps within categories such as GAME, TOOLS, or FAMILY to understand what makes them successful.
It enables category-wise performance analysis, helping companies benchmark competitors and improve their own apps.

```python
#Top Rated App in Each Category
query = """
SELECT *
FROM(
SELECT App,
Category,
Rating,
ROW_NUMBER() OVER(
PARTITION BY Category
ORDER BY Rating DESC
) rn
FROM apps
)
WHERE rn=1;
"""
pd.read_sql(query, conn)
```

This query helps businesses identify the best-performing app in each category.
Companies can analyze these top-rated apps to understand the features, quality, and user experience that contribute to high ratings.
The results can be used for competitive analysis, market research, and benchmarking when developing new apps.
